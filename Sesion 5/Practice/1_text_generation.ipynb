## **4. Generative pre-training Transformer - GPT**

![](../assets/gpt.png)

Los modelos tipo GPT, introducidos por Radfor, et.al., de OpenAI, al igual que los modelos BERT, hacen uso extensivo de la arquitectura de transformers como hemos estado viendo. Las diferencias claves se podrían resumir en:

1. GPT utiliza bloques de **Transformer Decoder** encadenados, mientras que el modelo BERT utiliza bloques de *Transformer Encoder*
2. GPT se centra en la generación de texto basado en un contexto, la tarea principal es la predicción del siguiente token en la secuencia, mientras que BERT se centra en el completado de partes de una secuencia, en función de un contexto anterior y posterior a la secuencia de entrada. Entonces BERT se centra en la construicción de representación de lenguage, mientras que GPT se centra en la generación de texto en función de un contexto.

Sin embargo, ambos se basan en la misma premisa de pre-entrenar el modelo en tareas no-supervisadas o semi-supervisadas para que el modelo aprenda las representaciones semánticas del lenguage y luego al modelo se le pueda hacer fine tuning a tareas posteriores.

## **Comentario sobre el Modelo GPT-2**
Este modelo se compone de un transformer causal con 12 capas (h0–h11) en esta versión base. Cada capa incluye:

- LayerNorm (normalización de capas) para estabilidad numérica.

- Self-Attention (attn) para capturar dependencias entre tokens.

- MLP (c_fc, c_proj) con activación NewGELUActivation para procesamiento no lineal.

El embedding de entrada (wte) y de posición (wpe) codifican tokens y posiciones.
La salida final del transformer se conecta a lm_head (una capa lineal) que proyecta al vocabulario de 50257 tokens.

## **Comentario sobre el Modelo GPT-Neo**

Este Modelo tiene 24 capas (h0–h23), el doble de la base de GPT-2, lo que permite modelar dependencias más complejas. Cada capa incluye:

## **8. Implementando una función de generación de texto en Inglés con GPT-Neo**

Ahora, la idea es que este modelo nos sirva para generar texto de forma recurrente e incremental. En la última capa de los modelos tipo GPT encontrarémos un tensor con forma $(b, s, v)$, donde:

- $b$: Es el tamaño del bache, o la cantidad de secuencias a procesar.
- $s$: Es la longitud de la secuencia de entrada.
- $v$: Es el tamaño del vocabulario del modelo, cuantos tokens soporta.

Pero este es el tensor de salida, por qué tiene la forma de la secuencia de entrada?, porque cada posición en la salida corresponde a la la predicción del siguiente token de esa posición en la secuencia de entrada. En otras palabras, lo que obtenemos como predicción, es una secuencia de igual tamaño a la de entrada, movida un token hacia adelante, lo que efectivamente nos predice un solo token a la vez y por ende, el token que nos insteresa, es el último.

Lo que obtenemos en este tensor es además los logits de TODO el vocabulario del modelo, con los cuales podemos calcular las probabilidades de que cada uno sea el que continue en la secuencia. Hay varias formas de decodificar el siguiente token, la más fácil de implementar sería una decodificación codiciosa (greedy) del siguiente token, que consiste simplemente en seleccionar el token con la probabilidad más alta. Este es un enfoque simple y efectivo para algunos casos, pero al mismo tiempo sufre de poca variabilidad e incluso puede caer en generación repetitiva.

Otra opción es el muestreo, ya que justamente podemos obtener probabilidades del siguiente token, lo más lógico sería muestrear con esas opciones de probabilidad, de este modo podemos obtener mayor diversidad a la hora de generar el texto, al costo eso si de que haya mayor aleatoridad ya que se le daría la oportunidad a incluso tokens con baja probabilidad, de ser seleccionados.

Otra opción podría ser un balanceo entre una decodificación greedy y una por muestreo, en función de otro hiperparámetro que podemos definir. Esta sería una técnica muy común en el contexto de Reinforcement Learning llamade e-greedy. Se hace la aclaración de que en este ejemplo no harémos nada de RL, solamente se hace mención de esta técnica para balancear entre explotación y exploración.

```python
def generate(
        model: nn.Module,  # Modelo de lenguaje causal (GPT-2, GPT-Neo, etc.)
        tokenizer: PreTrainedTokenizerBase,  # Tokenizador correspondiente al modelo
        start: str,  # Texto inicial para generar la secuencia
        max_length: int = 100,  # Longitud máxima de tokens a generar
        eps: float = 0.5,  # Parámetro e-greedy para balance entre greedy y muestreo
        top_n: int = 5,  # Número de mejores opciones a considerar para análisis
        return_iterations: bool = False,  # Si True, devuelve detalle de cada iteración
        device: str = "cpu") -> Tuple[str, Optional[pd.DataFrame]]:  # Dispositivo (CPU/GPU) y tipo de retorno

    output = [start]  # Inicializamos la salida con el texto inicial
    iterations = []  # Lista para guardar detalles de cada iteración (opcional)

    with torch.no_grad():  # Desactivamos gradientes, solo generamos texto
        input_ids = tokenizer_gptneo(output[-1], return_tensors='pt')['input_ids'].to(device)  # Tokenizamos y pasamos a GPU/CPU

        for _ in range(max_length):  # Iteramos hasta generar max_length tokens
            logits = model_gptneo(input_ids=input_ids).logits  # Predicción cruda (logits) del siguiente token
            probs = torch.softmax(logits[0, -1, :], dim=-1)  # Convertimos logits en probabilidades
            sorted_tokens = torch.argsort(probs, dim=-1, descending=True)  # Ordenamos tokens por probabilidad descendente

            # Política e-greedy: balance entre greedy (tomar el mejor token) y muestreo
            if np.random.random_sample(1)[0] < eps:
                next_token = sorted_tokens[0].unsqueeze(dim=0)  # Tomamos el token más probable (greedy)
            else:
                next_token = torch.multinomial(probs, 1)  # Muestreamos token según su distribución de probabilidad

            if return_iterations:  # Guardamos detalle de la iteración para análisis
                iteration = {'input': ''.join(output)}  # Texto generado hasta ahora
                best_n = sorted_tokens[:top_n].cpu().tolist()  # Top N tokens más probables
                # Creamos diccionario con token decodificado y su probabilidad
                choices = {f'Choice #{choice+1}': f'{tokenizer.decode(token)} ({prob:.4f})'
                           for choice, (token, prob) in enumerate(zip(best_n, probs[best_n].cpu().tolist()))}
                iteration.update(choices)  # Actualizamos iteración con las opciones
                iterations.append(iteration)  # Guardamos la iteración en la lista

            output.append(tokenizer.decode(next_token))  # Decodificamos el token y lo agregamos al output
            input_ids = torch.cat([input_ids, next_token.unsqueeze(dim=0)], dim=-1)  # Actualizamos input_ids para siguiente predicción

        output_text = ''.join(output)  # Convertimos lista de tokens en un string final

        if not return_iterations:
            return output_text, None  # Retornamos solo el texto
        else:
            df = pd.DataFrame(iterations)  # Creamos DataFrame con detalle de cada iteración
            return output_text, df  # Retornamos texto + DataFrame
```

### **8.1. Generación de Texto usando la Función definida en el Paso 8**

Vamos a usar la funcion *Generate* definida en el Paso 8, para generar texto usando el Modelo *GPT-Neo* y su Tokenizador (*tokenizer_gpteno*). Veamos como le va haciendo predicciones en la secuencia del texto.

#### **8.1.1. Uso de la Funcion con eps=1.0**

Veamos como le va generando texto usando un eps=1.0.

```python
# Generamos texto usando la función 'generate' con nuestro modelo GPT-Neo
# max_length=15 limita la generación a 15 tokens
# eps=1.0 significa que siempre tomaremos el token más probable (modo greedy)
# top_n=10 nos permite analizar las 10 opciones más probables por iteración
# return_iterations=True guarda información de cada paso en un DataFrame
output_text, iterations_df = generate(
    model_gptneo,        # Modelo GPT-Neo cargado previamente
    tokenizer_gptneo,    # Tokenizador correspondiente a GPT-Neo
    text,                # Texto inicial para comenzar la generación
    max_length=15,       # Número máximo de tokens a generar
    eps=1.0,             # Política greedy (siempre el token más probable)
    top_n=10,            # Número de opciones a mostrar por iteración
    return_iterations=True,  # Retornar detalle de cada iteración
    device=device        # Dispositivo (CPU o GPU) donde se realiza la generación
)

# Mostramos el texto generado completo
print("Este es el texto completo generado en Inglés por el Modelo GPT-Neo: \n", output_text)

# Mostramos las primeras 15 filas del DataFrame de iteraciones para analizar decisiones del modelo
print("\n Muestra de Texto generado e Iteraciones realizadas por el Modelo GPT-Neo")
iterations_df.head(15)
```

## **12. Preprocesamiento del dataset con tokenización**

En este parte del código definimos una función de preprocesamiento que se encarga de tokenizar los chistes en Inglés utilizando el modelo GPT-Neo. La función aplica truncamiento y padding para asegurar que todas las secuencias tengan la misma longitud máxima (max_len), lo cual es necesario para que el entrenamiento sea más eficiente y consistente.

```python
# En este bloque definimos una función de preprocesamiento para el dataset.
# La idea es tokenizar cada chiste utilizando el tokenizador de GPT-Neo.
#
# 1. La función externa `preprocess_function(max_len)` recibe como parámetro
#    la longitud máxima (`max_len`) que queremos para cada secuencia.
#
# 2. Dentro de ella, se define la función interna `_preprocess_function(examples)`,
#    la cual toma como entrada un conjunto de ejemplos del dataset.
#
# 3. Esta función aplica el tokenizador sobre la columna "Joke", truncando los
#    textos que sean demasiado largos y rellenando con padding hasta la longitud
#    máxima definida.
#
# Con esto garantizamos que todas las secuencias tengan la misma longitud,
# lo cual es un requisito indispensable para poder entrenar el modelo de forma
# eficiente en tareas de generación de texto.
def preprocess_function(max_len):
    def _preprocess_function(examples):
        return tokenizer_gptneo(
            examples['Joke'],
            max_length=max_len,
            truncation=True,
            padding='max_length'
        )
    return _preprocess_function
```

## **14. Tokenización del Dataset Reducido**

Tomamos el Dataset Reducido que contiene 20 mil Registros de chistes en inglés, que usaremos para entrenar un modelo GPT-Neo. Pero para hacerlo hay que Tokenizarlo. El codigo en la celda permite definir la longitud máxima de las secuencias y es clave este valor, que debe escoger acorde a los valores de longitud de palabra maxima, minima y media del conjunto de datos. Enseguida, aplicamos la tokenización de los chistes, y proseguimos con una división del conjunto de datos en dos subconjuntos: Entrenamiento y Prueba. Estos dos subconjuntos los adaptamos al formato de PyTorch. Y como final, configuramos un Data Collator adecuado para modelos causales, asegurando que el entrenamiento se realice de manera correcta.


```python
# Definimos la longitud máxima de las secuencias que procesará el modelo.
max_long = 64

# Aplicamos la función de preprocesamiento al conjunto de datos de entrenamiento.
# Esto tokeniza los chistes con la longitud máxima definida y los prepara para el modelo.
tokenized_dataset = dataset["train"].map(preprocess_function(max_long), batched=True)

# Dividimos el dataset en entrenamiento y prueba.
# Con test_size=0.1 reservamos el 10% de los datos para evaluar el rendimiento del modelo.
tokenized_dataset = tokenized_dataset.train_test_split(test_size=0.1)

# Ajustamos el formato de los datos para que solo contenga lo que el modelo necesita:
# los IDs de los tokens y la máscara de atención, en formato tensor de PyTorch.
tokenized_dataset.set_format(type="torch", columns=["input_ids", "attention_mask"])

# Creamos un Data Collator, que se encarga de organizar los lotes de datos durante el entrenamiento.
# En este caso es importante indicar mlm=False porque GPT-Neo es un modelo causal
# y no funciona con enmascarado de palabras como los modelos tipo BERT.
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer_gptneo,
    mlm=False   # GPT-Neo es causal LM, no Masked LM
)
```

## **15. Configuración del Entrenador con Hugging Face Trainer**

En esta sección definimos los parámetros globales de entrenamiento (épocas, tasa de aprendizaje, tamaño de lote, etc.) y configuramos el objeto `Trainer`.  
Este componente nos permitirá entrenar y evaluar el modelo GPT-Neo de manera estructurada, aprovechando las utilidades que ofrece la librería de Hugging Face para la gestión de datasets, métricas y checkpoints.


```python
# Definimos el tamaño de los lotes de entrenamiento y validación
batch_size = 8

# Calculamos cada cuántos pasos se mostrará el log, en función del tamaño del dataset
logging_steps = len(tokenized_dataset['train']) // batch_size

# Establecemos los parámetros globales de entrenamiento para Hugging Face Trainer
training_args = TrainingArguments(
    output_dir='./hf-gpt',              # Carpeta donde se guardarán los resultados y checkpoints
    overwrite_output_dir=True,          # Permite sobrescribir el directorio de salida si ya existe
    num_train_epochs=3,                 # Número de épocas de entrenamiento
    learning_rate=2e-5,                 # Tasa de aprendizaje inicial
    per_device_eval_batch_size=batch_size,   # Tamaño de lote para evaluación
    per_device_train_batch_size=batch_size,  # Tamaño de lote para entrenamiento
    weight_decay=0.01,                  # Factor de regularización para evitar sobreajuste
    evaluation_strategy='epoch',        # Evaluar el modelo al final de cada época
    save_strategy='epoch',              # Guardar checkpoints al final de cada época
    load_best_model_at_end=True,        # Cargar el mejor modelo tras el entrenamiento
    disable_tqdm=False,                 # Mantener la barra de progreso activa
    logging_steps=logging_steps,        # Frecuencia de registro de métricas
    logging_strategy='epoch',           # Registrar métricas cada 'logging_steps' pasos
    report_to='none'                   # No enviar métricas a plataformas externas
)

# Definimos el entrenador de Hugging Face,
# indicando el modelo, los parámetros, los datasets, el tokenizador y el collator
trainer = Trainer(
    model=model_gptneo,                     # El modelo GPT-Neo preentrenado que ajustaremos
    args=training_args,                     # Los parámetros globales definidos arriba
    train_dataset=tokenized_dataset['train'],   # Subconjunto de datos para entrenamiento
    eval_dataset=tokenized_dataset['test'],     # Subconjunto de datos para evaluación
    tokenizer=tokenizer_gptneo,             # Tokenizador correspondiente al modelo
    data_collator=data_collator             # Objeto que organiza los datos en lotes
)
```
